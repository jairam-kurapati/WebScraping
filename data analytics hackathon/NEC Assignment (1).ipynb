## Create a scatter plot for the prices in first page?

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("flipkart_mobiles_under_50000.csv")


first_page = df.head(24)


prices = (
    first_page["Prices"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)


plt.figure()
plt.scatter(range(len(prices)), prices)
plt.xlabel("Product Index (First Page)")
plt.ylabel("Price (INR)")
plt.title("Scatter Plot of Mobile Prices on First Page")
plt.show()
```

## Findout the standard deviation on price column in page3?

```python
import pandas as pd


df = pd.read_csv("flipkart_mobiles_under_50000.csv")


page_size = 24


start = (3 - 1) * page_size
end = start + page_size
page3 = df.iloc[start:end]


prices_page3 = (
    page3["Prices"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)


std_dev = prices_page3.std()

print("Standard Deviation of prices on Page 3:", std_dev)
```

## Find the difference between original price and selling price for top 30 products.(Use Rating feature for sorting)

```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")


df["Original_Price"] = (
    df["Original_Price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)

df["Selling_Price"] = (
    df["Selling_Price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)


top_30 = df.sort_values(by="Rating", ascending=False).head(30)


top_30["Price_Difference"] = (
    top_30["Original_Price"] - top_30["Selling_Price"]
)

print(top_30[[
    "Product Name",
    "Rating",
    "Original_Price",
    "Selling_Price",
    "Price_Difference"
]])
```

## Q1. For a given product category (e.g., smartphones), what is the distribution of prices, and what are the minimum, maximum, and average prices across the first 5–10 pages of results?


```python
import pandas as pd


df = pd.read_csv("flipkart_mobiles_under_50000.csv")


df["Price_numeric"] = (
    df["Prices"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)


min_price = df["Price_numeric"].min()
max_price = df["Price_numeric"].max()
avg_price = df["Price_numeric"].mean()


print("Minimum Price:", min_price)
print("Maximum Price:", max_price)
print("Average Price:", avg_price)
```

## Q2. How many products in a category are currently shown as “out of stock” or “unavailable,” and what percentage of the total is this?


```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")


unavailable_count = df[
    df["Availability"].str.contains("out of stock|unavailable", case=False, na=False)
].shape[0]


total_products = len(df)


percentage = (unavailable_count / total_products) * 100

print("Unavailable products:", unavailable_count)
print("Percentage unavailable:", percentage)
```

## Q3. For one brand and category, how often do discounts appear (e.g., percentage of products with a discount tag vs. without)?


```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")


brand_df = df[df["Brand"] == "Samsung"]


discounted = brand_df[brand_df["MRP"] > brand_df["Selling_Price"]]


total = len(brand_df)
discount_count = len(discounted)
discount_percentage = (discount_count / total) * 100

print("Total products:", total)
print("Products with discount:", discount_count)
print("Discount frequency (%):", discount_percentage)
```

## Q4. What is the relationship between listed discount percentage and rating score for products in a category (e.g., do higher discounts tend to have lower ratings)?


```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")


df["Discount_Percentage"] = (
    (df["MRP"] - df["Selling_Price"]) / df["MRP"]
) * 100


correlation = df["Discount_Percentage"].corr(df["Rating"])

print("Correlation between discount and rating:", correlation)
```

## Q5. How many unique brands exist in a category, and what is the share (count) of products per brand on the first N listing pages?


```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")


N = 5
page_size = 24
df_n_pages = df.head(N * page_size)


unique_brands = df_n_pages["Brand"].nunique()


brand_counts = df_n_pages["Brand"].value_counts()

print("Number of unique brands:", unique_brands)
print("\nProduct count per brand:")
print(brand_counts)
```

## Q6. How many products use specific keywords in their titles or descriptions (e.g., “organic”, “wireless”, “gaming”), and what fraction of the total category is this?


```python
import pandas as pd


df = pd.read_csv("flipkart_mobiles_under_50000.csv")


df["combined_text"] = (
    df["Product Name"].fillna("") + " " + df["Description"].fillna("")
).str.lower()


keywords = ["gaming", "wireless", "organic"]

total_products = len(df)

for kw in keywords:
    count = df["combined_text"].str.contains(kw, na=False).sum()
    fraction = (count / total_products) * 100
    print(f"{kw}: {count} products ({fraction:.2f}%)")
```

## Q7. Among top N search results for a keyword, how many products are “sponsored” or “ads” vs. organic results, if the site marks them differently?


```python
import pandas as pd

df = pd.read_csv("flipkart_mobiles_under_50000.csv")

N = 20
top_n = df.head(N)

sponsored_count = top_n[top_n["Sponsored"] == True].shape[0]
organic_count = top_n[top_n["Sponsored"] == False].shape[0]

total = len(top_n)

print("Sponsored products:", sponsored_count)
print("Organic products:", organic_count)
print("Sponsored %:", (sponsored_count / total) * 100)
print("Organic %:", (organic_count / total) * 100)
```
